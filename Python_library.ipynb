###list
list is **mutable, ordered** collection of any types of data

```python
# list is written in [] square brackets, list can store any data types
lt = [2,4.7,3,"hii",True]
print(lt)
print(type(lt))
type(lt)
```

```python
# Mutable (changable)
lt = [2,4.7,3,"hii",True]
lt.append(4545)
print(lt)
# Indexing in list
print(lt[4])
# Slicing is also possible
print(lt[1:4])   # print 1 to 4-1 for [1:4]
```

```python
# Allow Duplicate value in list
lt1 = [2,3,2,4,56,6,4,3,6,3,3,3,3]
print(lt1)
```

```python
# Empty list
lt_2 = []
print(lt_2)  #[]
lt_2.append(99)   #[99]
print(lt_2)

# Nested list
lt_3 = [2,[3,4,[8,"hi"],9],True,[]]
print(lt_3)    #[2, [3, 4, [8, 'hi'], 9], True, []]

# Update existing elements
lt_3[1] = 50
print(lt_3)    # [2, 50, True, []]
lt_3.insert(3,888)    # insert any element(888) at any index(3)
print(lt_3)       # [2, 50, True, 888, []]

# extend many elements in list
lt_4 = [1,]
lt_4.extend(["x",6,"hiii"])
print(lt_4)   # [1, 'x', 6, 'hiii']

# remove selected elements
lt_4.remove(6)
print(lt_4)   # [1, 'x', 'hiii']

# pop {remove ele by index no}
lt_4.pop(0)
print(lt_4)    # ['x', 'hiii']

# clear [remove all ele from list]
lt_4.clear()
print(lt_4)    # []

# index [give the index of ele]
lt_5 = [2,22,33,45,"hi"]
lt_5.index(45)   # 3

# count {count the selected ele}
lt_5.count(22)  # 1
```

```python
# sort()--- for sorting ele of list and reverse()--- for reversing ele of list
lt_6 = [2,6,7,3,2,9,3,]
lt_6.sort()
print(lt_6)   # [2, 2, 3, 3, 6, 7, 9]
lt_6.reverse()
print(lt_6)   # [9, 7, 6, 3, 3, 2, 2]

# join the lists
lt_7 = lt_6 + lt_5
print(lt_7)   # [9, 7, 6, 3, 3, 2, 2, 2, 22, 33, 45, 'hi']

# length of list
print(len(lt_7))   # 12
```

###set
set is an **unordered, unindexed, mutable** collection of unique elements

```python
# written inside {} curly braces and with set() functions
st = set()    # empty set
print(st)
print(type(st))

# with set() functions
st1 = set([2,4,3,5,6,7,6,7])
print(st1)
type(st1)

# no duplicate in sets   (unindexed)
st_1 = {3,5,8,8,3,9,5,5,5,5,5}
print(st_1)     # {8, 9, 3, 5} will be in any order (unordered)
type(st_1)
```

```python
# Functions of sets
# add ele in set
st_2 =set()
st_2.add(6)
print(st_2)

# add multiple elements in set
st_2.update([7,3,2,9,88])
print(st_2)

# remove ele from set [existing element] if not exist then return error
st_2.remove(2)
print(st_2)

# remove element that is not exist in set  [return original set if not exist nut no error]
st_2.discard(66)
print(st_2)

# remove random ele from set
st_2.pop()
print(st_2)

# clear all ele from set
st_2.clear()
print(st_2)   # set() -- empty set

# copy -- to make a duplicate set
st_21 = {7,5,4,2,8,9,0,33,45}
st_22 = st_21.copy()
print(st_22)    # {0, 33, 2, 4, 5, 7, 8, 9, 45}
```

```python
# operations of set { just like mathematics -- union, intersection, diff. etc}
# union -- addition of sets
a = {1,2,3,4,5}
b = {4,5,7,3,8}
print(a.union(b))     # {1, 2, 3, 4, 5, 7, 8}
print(a|b)            # {1, 2, 3, 4, 5, 7, 8}

# intersection -- commom element from both sets
print(a.intersection(b))    # {3, 4, 5}
print(a&b)                  # {3, 4, 5}

# difference --- substraction of a & b
print(a.difference(b))    # {1, 2}
print(a-b)                # {1, 2}
print(b-a)                # {8, 7}

# symmetric difference ---
print(a.symmetric_difference(b))    # {1, 2, 7, 8}
print(a^b)                          # {1, 2, 7, 8}

# others -- [subset, superset, disjoint]
c = {1,2,3,4,5,4,5,7,3,8}
d = {4,5,7,3,8}
print(c.issubset(d))    # False
print(c.issuperset(d))  # True
print(d.isdisjoint(c))  # False
```

###Dictionary
dictionary is **unordered, mutable, indexed** coleection of **KEY-VALUE**

```python
# Enclosed in curly braces {}
info = {
    "name":"chandan kumar",
    "age" : 20,                # "age" is KEY & 20 is VALUE
    "age":20,                  # no duplicate key-value allowed
    "height" : 165
}
print(info)
is_adult = info.get("age")
print(info["age"])        # 20
print(is_adult)           # 20
print(type(info))         # <class 'dict'>
print(info.get("name"))   # chandan kumar

# replace value
info["age"] = 21          # 20 is updated now as 21
print(info["age"])        # 21

# add new KEY and VALUE in dictionary
info["weight"] = 60
print(info)                # weight will added in info

# type of key
type(info["weight"])        # int

# null dictionary
dct = {}
print(dct)       # {}
```

```python
# Nested dictionary
student = {
    "Name": "chandan",
    "Surnamename": "kumar",
    "age":20,
    "height":164,
    "subject" :{
        "phy":99,
        "chem":95,
        "math":98
    },
    "Adult": True
}
print(student)
print(student["subject"]["math"])
print(student["subject"])
```

```python
# functions of dictionary
dct ={}
print(dct)

# add key-value in dict
dct["city"]="durgapur"
dct["state"]="west bengal"
dct["country"]="india"
print(dct)

# update key-value in dict
dct["city"]="kolkata"
print(dct)

# remove key-value
dct.pop("state")
print(dct)

# remove last inserted key-value
print(dct.popitem())

# copy to duplicate
dct_1 = dct.copy()
print(dct_1)

# clear all key-value
dct.clear()
print(dct)
```

```python
# print all key from dict
print(student.keys())

# print all values from dict
print(student.values())

# print all key-value pair
print(student.items())

# get value
print(student.get("age"))   # if exist then return value
print(student.get("city"))  # if not exist then return None

# conversion
print(list(student))    #convert to list
print(tuple(student))   #convert to tuple
print(set(student))     #convert to set
```

###Operators

```python
1 # Arithemetic operators
num_1 = 30
num_2 = 50

#addition
sum = num_1 + num_2
print("sum:",sum)

#substration
diff = num_2 - num_1
print("diff:",diff)

#multiplication
mul = num_1 * num_2
print("mul:",mul)

#division
div = num_2 / num_1
print("div:",div)

# modulus (remainder)
mod = num_2 % num_1
print("mod:",mod)

# exponent
exp = num_1 ** num_2
print("exp:",exp)    # 30 power 50
```

```python
2 # Assignment operators
aa = 5
print(aa)   # 5

# + equal
aa += 5
print(aa)   # (5+5 = 10)

# - equal
aa -= 5
print(aa)   # (10-5 = 5)

# *=
aa *= 5
print(aa)   # (5*5 = 25)

# /=
aa /= 5
print(aa)   # (25/5 = 5.0)

# %=
aa %= 5
print(aa)   # (5%5 = 0.0)

# //=
aa //= 5
print(aa)   # (0//5 = 0.0)
```

```python
3 # Comparison operator
a1 = 5
b1= 10
print(a1==b1)  # check equal to
print(a1!=b1)  # check not equal to
print(a1 > b1)   # check greater than
print(a1 < b1)   # check less than
print(a1 >= b1)  # check greater than or equal to
print(a1 <= b1)  # check less than or equal to
```

```python
4 # Logical operator
a2 = 10
print(a2 > 20 and a2 > 5)    # both statement must be correct
print(a2 > 20 or a2 > 5)     # any one will correct
print(not(a2 > 2 and a2 > 5))   # reverse the value---statement is true but false
print(not(a2 > 20 or a2 > 50))    # reverse the value--- statement is false but true
```

```python
5 # Identity operator
a3 = 5
b3 = 5
b2 = 50
print(a3 is b3)     # -- True   both are same
print(a3 is b2)     #-- False   both aren't same
print(a3 is not b2)   # -- True   both are not same it is correct so True
```

```python
6 # Membership operator
a4 = 5
b4 = 10
c = [1,2,3,4,5]
print(a4 in c)    # True
print(b4 in c)     # False
print(a4 not in c)   # False
print(b4 not in c)    # True
```

```python
# creating a nested pie chart
from matplotlib import pyplot as plt
import numpy as np
# Creating dataset
size = 6
cars = ['AUDI', 'BMW', 'FORD',
        'TESLA', 'JAGUAR', 'MERCEDES']

data = np.array([[23, 16], [17, 23],
                 [35, 11], [29, 33],
                 [12, 27], [41, 42]])

# normalizing data to 2 pi
norm = data / np.sum(data)*2 * np.pi

# obtaining ordinates of bar edges
left = np.cumsum(np.append(0,
                           norm.flatten()[:-1])).reshape(data.shape)

# Creating color scale
cmap = plt.get_cmap("tab20c")
outer_colors = cmap(np.arange(6)*4)
inner_colors = cmap(np.array([1, 2, 5, 6, 9,
                              10, 12, 13, 15,
                              17, 18, 20]))

# Creating plot
fig, ax = plt.subplots(figsize=(10, 7),
                       subplot_kw=dict(polar=True))

ax.bar(x=left[:, 0],
       width=norm.sum(axis=1),
       bottom=1-size,
       height=size,
       color=outer_colors,
       edgecolor='w',
       linewidth=1,
       align="edge")

ax.bar(x=left.flatten(),
       width=norm.flatten(),
       bottom=1-2 * size,
       height=size,
       color=inner_colors,
       edgecolor='w',
       linewidth=1,
       align="edge")

ax.set(title="Nested pie chart")
ax.set_axis_off()

# show plot
plt.show()
```

```python
# 3D surface plot
x = np.outer(np.linspace(-2, 2, 10), np.ones(10))
y = x.copy().T
z = np.cos(x**2 + y**3)

fig = plt.figure()
ax = plt.axes(projection='3d')   # 3d projection

ax.plot_surface(x, y, z, cmap='viridis', edgecolor='green')
plt.title('3D surface plot')
plt.show()
```

```python
# Wireframe plot

def f(x, y):
    return np.sin(np.sqrt(x**2 + y**2))

x = np.linspace(-1, 5, 10)
y = np.linspace(-1, 5, 10)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

fig = plt.figure()
ax = plt.axes(projection='3d')
ax.plot_wireframe(X, Y, Z, color='green')
ax.set_title('Wireframe Plot')
plt.show()
```

```python
# 3D contour plot

def fun(x, y):
    return np.sin(np.sqrt(x**2 + y**2))

x = np.linspace(-10, 10, 40)
y = np.linspace(-10, 10, 40)
X, Y = np.meshgrid(x, y)
Z = fun(X, Y)

fig = plt.figure(figsize=(8, 6))   # size of figure
ax = plt.axes(projection='3d')
ax.plot_surface(X, Y, Z, cmap='cool', alpha=0.8)

ax.set_title('3D Contour Plot')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')

plt.show()
```

* Stemming-- the process of reducing a word to its Root word

ex:- actor,actress,acting---> act

```python
port_stem = PorterStemmer()
```

```python
def stemming(content):
  stemmed_content = re.sub('[^a-zA-Z]',' ',content)
  stemmed_content = stemmed_content.lower()
  stemmed_content = stemmed_content.split()
  stemmed_content = [port_stem.stem(word) for word in stemmed_content if not word in stopwords.words('english')]
  stemmed_content = ' '.join(stemmed_content)
  return stemmed_content
```

```python
import nltk
nltk.download('stopwords')
news_dataset['content'] = news_dataset['content'].apply(stemming)  # applying stemming to contents
```

    0 --> spam
    
    1 --> ham

```python
# accuracy_score on test data
x_test_predicted = model.predict(x_test_features)
x_test_predicted_accuracy = accuracy_score(x_test_predicted, y_test)
print(x_test_predicted_accuracy)
```

# Building Linear Regression from scratch

1. importing dependencies

```python
# importing dependencies
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

step:
1. set learning rate & no of iterations; initialize random weight and bias value
2. build linear regression equation (y= wx + b)
3. finnd the y pred value for given x value for the corresponding weight & bais.
4. check the loss functuionn for these parameters values (diff b/w y pred and true y)
5. udate the parameters values using gradiet descent.
6. step 3, 4, 5 are repeated till we get minimum loss function
7. finally, we will get the best model (best weight and bias value) as it has minimum loss function.

```python
class Linear_Regression():  #class

  # initiating the paarameters(learning_rate and no of iterations is hyperparameters)
  def __init__(self, learning_rate, no_of_iterations):  # function/method
    self.learning_rate = learning_rate
    self.no_of_iterations = no_of_iterations

  def fit(self, x, y ):    # function/method
    self.x = x       # x
    self.y = y       # y
    # m--> no. of training examples & n --> no. of features
    self.m, self.n = x.shape  # m--> rows and n--> columns

  # initiating the weight and bias
    self.w = np.zeros(self.n) # weight
    self.b = 0       # bias


    # implementing gradient descent
    for i in range(self.no_of_iterations):
      self.update_weights()   # upadte_weight each iterations

  def update_weights(self):
    y_prediction = self.predict(self.x)

    # calculate the gradient using fromulas
    dw = - (2 * (self.x.T).dot(self.y - y_prediction)) / self.m

    db = - (2 * np.sum(self.y - y_prediction)) / self.m

    # updating the weight using formulas
    self.w = self.w - self.learning_rate*dw  # weight = w - alpha*dw, where alpha =learning_rate
    self.b = self.b - self.learning_rate*db  # bias = b - alpha*db, where alpha = learning_rate

  def predict(self, x):

    return x.dot(self.w) + self.b  # y = mx + c (y=wx+b)
```

# Building Logistic Regression from Scratch

 1. Sigmoid function :
*    y' = 1/(1+ e`-z), where z = w.x + b
***
 2. gradient descent:\
*   w2 = w1 - L*dw
*   b1 = b1 - L*db\
where w-> weight, b-> bias, m-> no of data point

*   dw = 1/m*(y' - y)x
*   db = 1/m*(y' - y)

dw--> partial derivative of cost
function w.r.t w\ # how much cost function decrease wrt w,
db-->  partial derivative of cost function w.r.t b # how much cost function decrease wrt b,
***
3. Loss function :
 binary cross entropy loss function or log loss formula\
*    L(y,y') = -(y log y')+(1-y)log(1-y')\

4.cost function :
 mean of loss function,
*   j(w,b) = 1/m E(L(y, y')) = -1/m E(y, log y' + (1+y)log(1-y'))\
where, E--> sumition sign.

```python
# LOGISTIC REGRESSION from scratch

import pandas as pd
import numpy as np
```

Work flow / step for building Logistic Regression:\
1. set learning rate & no of iterations; initialize random weight and bias value.\
2. build Logistic Regression function.( Segmoid Function )\
3. update the parameters using gradient descent\
finally, we get the best model ( best weight and bias value) as it has minimum cost function.\
4. Build the "predict" function to determine the class of the data point.

```python
class Logistic_Regression():
# 1
  # declearing the laerning_rate & no of iterations (hyperparameters)
  def __init__(self, learning_rate, no_of_iterations):  # function/ methods
    self.learning_rate = learning_rate
    self.no_of_iterations = no_of_iterations
# 2
  def fit(self, x, y):
    self.x = x   # features
    self.y = y   # targets
  # m --> no of data point ( rows ),
  # and n --> no of input features ( columns )
    self.m, self.n = x.shape  # m--> no. of training examples & n --> no. of features

  # initiating the weight and bias value
    self.w = np.zeros(self.n)  # weight
    self.b = 0                 # bias
  # implementing gradient descent
    for i in range(self.no_of_iterations):
      self.update_weight()
# 3
  def update_weight(self):
  # y cap formula ( sigmoid function )
    z = self.x.dot(self.w) + self.b # z = wx + b
    y_cap = 1 / (1 + np.exp(-z))  # y' = 1/(1+e`power(-z))

  # gradient (derivative of both weight and bias)
    dw = (1/self.m) * np.dot(self.x.T, (y_cap - self.y)) # dw = 1/m*(y' - y)x
    db = (1/self.m) * np.sum(y_cap - self.y)  # db = 1/m*(y' - y)

    # updating weight & bias using gradient descent
    self.w = self.w - self.learning_rate * dw  # w2 = w1 - L*dw
    self.b = self.b - self.learning_rate * db  # b2 = b1 - L*db
   # Sigmoid Equation & Decision Boundary
# 4
  def predict(self, x_input): # Renamed parameter to x_input to avoid confusion

    z = x_input.dot(self.w) + self.b # Use parameter x_input, not self.x
    y_pred = 1 / (1 + np.exp( -z ))
    y_pred = np.where(y_pred > 0.5 , 1 ,0)  # if y_pred > 0.5 then 1 else 0
    return y_pred
```

#Building SVM classifier from scratch

ADV :- Work well with binary dataset, high dimension data, and efficient where margin of sepration is clear\
  * yi`(Yi_cap) : a. -1, if wTx1 + b =< -1,\
  b. 1, if wTx1 + b => 1
1. Equation of hyperplane :- W'Tx + b = Label ( value of w'Tx+b will be same values as predicted)
***
2. Kernals in SVM:
*  Linear Kernal --> K(x1,x2) = (x1.Tx2), use when features is more, if accuracy is not good then go with anothers kernal
*   Polynomial kernal --> K(x1,x2) = (x1Tx2 +r)'d
*   Radial Basis function(rdf) -->
K(x1,x2) = exp(-'y'.||x1-x2||'2)
*   Sigmoid kernal --> K(x1,x2) = tanh('y'.x1Tx2 + r), where'y' --> fcylen
***
3. Loss Function: diff bw predicted and actual values\
LOSS = 1/n(E'n, i=1)(Yi - Yi`)\
4. Hinge Loss = max(0, 1 - yi(wT xi +b)), where 0 --> for correct classification (if predicted and actual is same )and 1 --> for wrong classification (if predicted and actual is not same)
***
5. Gradient descent:\
* w2 = w1 - L*(dj/dw)
* b2 = b2 - L*(dj/db)
* if (yi. (w.x +b) >= 1): then,
dj/dw = 2lembda.w and dj/db = 0
* else (yi.(w.x +b)< 1): then, dj/dw = 2lambda.w -yi.xi and dj/db = yi



1. importing dependencies

```python
import numpy as np
```

```python
class SVM_Classifier():
 # initiating the hyperparameters
# 1
  def __init__(self, learning_rate, no_of_iterations, lambda_parameter):
    self.learning_rate = learning_rate
    self.no_of_iterations = no_of_iterations
    self.lambda_parameter = lambda_parameter

 # fitting the dataset to SVM classifier
# 2
  def fit(self, x, y):
    self.x = x
    self.y = y
    # m --> no. of datapoint/ no of rows
    # n --> no. of input_features/ no of columns
    # x --> shape of dataset
    self.m, self.n = x.shape

  # initiating the weight and bias value
    self.w = np.zeros(self.n)   # weight values
    self.b = 0                  # bias values
  # implementing gradient descent
  # for updating weight  with each iterations
    for i in range(self.no_of_iterations):
      self.update_weight()

 # function for updating the weight and bias
# 3
  def update_weight(self):
    # label encoding
    y_label = np.where(self.y <= 0, -1, 1) # if value is 0 take -1 else 1 in y

  # gradient (dw, db)
    for index, x_i in enumerate(self.x):  # index --> index value and x_i --> All feature values in one rows
      condition = y_label[index] *(np.dot(x_i, self.w) - self.b) >= 1
      if (condition == True):
        dw = 2 * self.lambda_parameter * self.w
        db = 0
      else:
        dw = 2 * self.lambda_parameter * self.w - np.dot(x_i, y_label[index])
        db = y_label[index]
      self.w = self.w - self.learning_rate * dw # update weight
      self.b = self.b - self.learning_rate * db  # update bias

 # predict the label for a given input value
# 4
  def predict(self, x):

    output = np.dot(x, self.w) - self.b   # y = wx + b --> hyperplane
    predicted_labels = np.sign(output)  # sidn function will return sign of value if 1 -> +1, and -1 as -1

    y_hat = np.where(predicted_labels <= -1, 0, 1)  # reversing the y_label
    return y_hat
```

# Building LASSO REGRESSION from scratch

LASSO --> Least Absolute Shrinkage and Selection Operator
* Supervised learning
* Rregression model
* Implement regularization(L1) to avoid Overfitting
* Build top of linear Regression

REGULARIZATION : Use to reduce the overfitting of the model by adding a penality term (lembda) to the model, LASSO use L1 regularization technique. WORKING: penalty reduces the value of coefficient(weight) to zero or eliminate few coefficients, so the model has fewer coefficient. As result overfittinh=g can be avoided.


1. Equation : y = wx + b
2. COST FUNCTION :\
j = 1/m[E'm,i=1(yi - yi_cap)'2 + lembdaE'n,j=1 Wj],
where
* m--> total no of data points
* n--> total no of input features
* yi--> Actual/true value
* yi_cap--> Predicted value
* lambda--> penaltyt term
* w--> parameter of the model
***
3. Gradient Descent:
* w2 = w1 - L *dj/dw, and
* b2 = b1 - L *dj/db,  L--> learning_rate
dj/dw and dj/db is the derivative of cost function w.r.t weight and bias.
* if (wj > 0):--> dj/dw = -2/m[[E'm,i=1 xj(yi - yi_cap)] + lembda]
* else (wj <= 0):--> dj/dw = -2/m[[E'm,i=1 xj(yi - yi_cap)] - lembda]
* and dj/db = -2/m[E'm,i=1(yi - yi_cap)]  for both cases  

```python
import numpy as np
```

```python
class Lasso_Regression():
# 1 initiating hyper-parameters
  def __init__(self, learning_rate, no_of_iterations, lambda_parameter):
    self.learning_rate = learning_rate
    self.no_of_iterations = no_of_iterations
    self.lambda_parameter = lambda_parameter

# 2 fit the data
  def fit(self, x, y):
    self.x = x
    self.y = y
  # m --> no of rows/datapoints in dataset and
  # n --> no of columns/features
    self.m, self.n = x.shape
  # initiating the weight and bias
    self.w = np.zeros(self.n)
    self.b = 0
  # implementing Gradient Descent algo for optimization
  # updating weight with each iterations
    for i in range(self.no_of_iterations):
      self.update_weight()

# 3 update the weight
  def update_weight(self):
  # linear equation of the model
    y_cap = self.predict(self.x)

  # Gradient (dw, db)
    dw = np.zeros(self.n)

    for i in range(self.n):
      if self.w[i] > 0:   # i --> represent each columns in dataset
        dw[i] = (-(2*(self.x[:,i]).dot(self.y - y_cap)) + self.lambda_parameter) / self.m  # + lembda
      else:
        dw[i] = (-(2*(self.x[:,i]).dot(self.y - y_cap)) - self.lambda_parameter) / self.m  # - lembda

    db = - 2 * np.sum(self.y - y_cap) / self.m  # for both cases

  # updating weight & bias
    self.w = self.w - self.learning_rate * dw  # w2 = w1 - L *dw
    self.b = self.b - self.learning_rate * db  # b2 = b1 - L *db

# 4 predict the target
  def predict(self, x):
    return self.x.dot(self.w) + self.b  # y = wx + b
```

# 5. Precision, Recall, &  F1 Score

**1. Precision** : it is  the ratio of **True Positive** to the **Total No of Predicted Positive**, its measure out of total +ve how much are actually +ve; also measures the *Error* caused by **False Positive**\

* it is good evalution metric when **FP** predcitions are critical.
***
     Precision = TP/(TP + FP)or, Total Predicted Positive
***
**2. Recall** : it is the ratio of **True Positive** to the **Total No of Actual Positive**, it measures, out of the total actual positive, how many are predicted as TP; also measures the *Error* caused by False -ve.

* Good when **FN** prdictions are critical
***
    Recall = TP/(TP + FN)or, Total Actual Positive
**3. F1-Score**: it is an important Binary classifications that combines **Precision & Recall**. F1-score is the Harmonic Mean of Precision & Recall.
* Useful when dataset has imbalanced classes.
***
    F1-Score = 2 * { (Precision * Recall) / (Precison + Recall) }

```python
# 1. importing the dependencies

# libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# K-means cluster
from sklearn.cluster import KMeans
```

2. Data Collection & Data proccessing

```python
mall_customers = pd.read_csv('/content/Mall_Customers.csv')
mall_customers
```

```python
mall_customers.info()
```

```python
# choosing the annual income & spending score columns in x variables
x = mall_customers.iloc[:, [3,4]].values  # annual income & spending score columns
y = mall_customers.iloc[:, 3].values  # annual income
```

```python
print(x)
```

```python
print(y)
```

3. choosing the correct no of clusters

WCSS - Within Cluster Sum of Squares

```python
# finding WCSS for different no of clusters

wcss = []  # empty list name as wcss

for i in range(1, 11): # for loop to run value 1 to 10

  kmeans = KMeans(n_clusters = i, init = 'k-means++', random_state = 42) # loading KMeans
  # n_clusters = no of cluster
  # init = no of times kmeans run to get best clustering result
  # random_state = ensure result will reproducible

  kmeans.fit(x) # fit data to model

  wcss.append(kmeans.inertia_)  # inertia = clusters value in each iterations
```

```python
print(wcss)
```

 Visualizing the elbow graph


```python
# plot an elbow graph with wcss and no of clusters
sns.set()

plt.plot(range(1,11), wcss)
plt.title('The Elbow point graph')
plt.xlabel('no. of clusters')
plt.ylabel('wcss')
plt.show()
```

The Elbow Point is 3, 5 (only 3 & 5 has form an sharp elbow )
* optimum no of clusters = 5

Training the k-means clustering model

```python
# training the k-means clustering model with cluster = 5
kmeans = KMeans(n_clusters = 5, init='k-means++', random_state = 0)

# return a label for each data points
y = kmeans.fit_predict(x)
print(y)
```

visualizing all the clusters

```python
# plotting all the clusters and their centroids

plt.figure(figsize=(8, 8))
#  -- x -- annual income
#  -- y -- spending score
plt.scatter(x[y==0, 0], x[y==0, 1], s = 50, c = 'green', label = 'Cluster 1' )
plt.scatter(x[y==1, 0], x[y==1, 1], s = 50, c = 'red', label = 'Cluster 2' )
plt.scatter(x[y==2, 0], x[y==2, 1], s = 50, c = 'yellow', label = 'Cluster 3' )
plt.scatter(x[y==3, 0], x[y==3, 1], s = 50, c = 'violet', label = 'Cluster 4' )
plt.scatter(x[y==4, 0], x[y==4, 1], s = 50, c = 'blue', label = 'Cluster 5' )
# plotting the centroids
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s = 100, color = 'black', label = 'Centroids')
plt.title('Customer Groups')
plt.xlabel('Annual Income')
plt.ylabel('Spending Score')
plt.legend()
plt.show()
```

Building the Neural Networks

```python
# 1 setting up the layers of the Neural Network

model = keras.Sequential([
    keras.layers.Flatten(input_shape = (28, 28)),# input_shape--> writting the matrix in single dimension or image dimension, (here 28, 28 but in rgb we use 28, 28, 3)
    keras.layers.Dense(50, activation = 'relu'),# hidden layers, activation = 'relu'  --> for accuracy
    keras.layers.Dense(50, activation = 'relu'),# hidden layers,
    keras.layers.Dense(10, activation = 'sigmoid')# 10 (0-9) for output neurons
])
```

model_2 with 3 hidden layers


* Sequential → a simple way to build a neural network layer by layer in order
* Input → defines the shape of your data coming in
* Flatten → converts multi-dimensional data into a 1D vector or Converts the 2D input (28 × 28) into a 1D vector
* Dense → a fully connected neural network layer

```python
# 2 setting up the layers of the Neural Network

from keras.models import Sequential
from keras.layers import Input, Flatten, Dense

model_2 = Sequential([
    Input(shape=(28, 28)),
    Flatten(),
    Dense(100, activation = 'relu'),
    Dense(100, activation = 'relu'),
    Dense(100, activation = 'relu'),
    Dense(10, activation = 'softmax')
])
# 100 neurons
# Each neuron connects to all 28 x 28 (784) inputs from Flatten
# Stands for Rectified Linear Unit : Helps the model learn non-linear patterns
```

Explaination of upper code :
1. First Dense layer
* Dense(100, activation='relu')
What it does:
Takes 784 inputs (from Flatten)
Learns 100 features
Why ReLU?
Keeps only positive values
Helps avoid vanishing gradients.

👉 Think of it as:
“Extract basic patterns from raw pixel data”
2. Second Dense layer
* Dense(100, activation='relu')
What it does:
Takes previous 100 outputs
Learns more abstract features.

👉 Now the network starts combining patterns:
edges → shapes → meaningful structures.

3. Third Dense layer
* Dense(100, activation='sigmoid')
What it does:
Still 100 neurons
Uses sigmoid activation
Problem:
Sigmoid squashes values between 0 and 1
Can cause vanishing gradients in deeper networks

👉 In practice:
Most modern models do NOT use sigmoid in hidden layers
ReLU (or variants) is usually better

4. Output layer
* Dense(10, activation='sigmoid')
What it does:
Produces 10 outputs (likely for 10 classes)
⚠️ Important issue:

Using sigmoid here depends on your task:

✅ If it's:
Multi-label classification (multiple answers possible)
→ sigmoid is OK
❌ If it's:
Multi-class classification (like digits 0–9)
→ this is WRONG

👉 You should use:
* Dense(10, activation='softmax')

Because:
Softmax ensures probabilities sum to 1
Picks one class properly

## Seprating features & labels

**creating the labels for cats & dogs** (Labels)
* cats --- 0
* dogs --- 1

```python
# creating a loop for labeling the cats & dogs

labels = []

for img_lbl in os.listdir('/content/Resized images'):

  name = img_lbl[0:3]# slicing by index 0 to 3 ( cat or dog )
  if name == 'dog':
    labels.append(1)
  else:
    labels.append(0)

print(labels)
```

```python
print(len(labels))
```

```python
# counting the no of dogs and cats

values, counts = np.unique(labels, return_counts = True)
print(values)
print("No of Cats (0):",counts[0], "NO of Dogs (1):", counts[1])
```
